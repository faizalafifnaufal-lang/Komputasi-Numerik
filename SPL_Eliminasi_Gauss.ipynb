```python
# Cell 1: Masukkan koefisien persamaan dan bentuk matriks augmented [A | b]
# Representasi sistem persamaan:
# a11*x + a12*y + a13*z = b1
# a21*x + a22*y + a23*z = b2
# a31*x + a32*y + a33*z = b3

# Contoh data uji:
# 2x + y - z = 8
# -3x - y + 2z = -11
# -2x + y + 2z = -3

A = [
    [2.0, 1.0, -1.0, 8.0],
    [-3.0, -1.0, 2.0, -11.0],
    [-2.0, 1.0, 2.0, -3.0]
]

n = len(A)

print("Matriks Augmented Awal [A | b]:")
for baris in A:
    print([f"{val:6.2f}" for val in baris])
```

```python
# Cell 2: Lakukan eliminasi Gauss untuk memperoleh matriks segitiga atas
for i in range(n):
    # Cek nilai pivot (jika 0, lakukan pertukaran baris sederhana)
    if A[i][i] == 0.0:
        for k in range(i + 1, n):
            if A[k][i] != 0.0:
                A[i], A[k] = A[k], A[i]
                break

    # Ubah elemen di bawah pivot menjadi nol (Eliminasi maju)
    for j in range(i + 1, n):
        faktor = A[j][i] / A[i][i]
        for k in range(i, n + 1):
            A[j][k] = A[j][k] - faktor * A[i][k]

print("Matriks Segitiga Atas (Hasil Eliminasi Gauss):")
for baris in A:
    print([f"{val:6.2f}" for val in baris])
```

```python
# Cell 3: Lakukan substitusi mundur untuk menghitung nilai z, y, dan x
x = [0.0] * n

# Dari baris terakhir ke baris pertama (indeks 2 ke 0)
for i in range(n - 1, -1, -1):
    x[i] = A[i][n]
    for j in range(i + 1, n):
        x[i] = x[i] - A[i][j] * x[j]
    x[i] = x[i] / A[i][i]

# Tampilkan nilai x, y, dan z
print("Penyelesaian SPLTV:")
print(f"Nilai x = {x[0]:.4f}")
print(f"Nilai y = {x[1]:.4f}")
print(f"Nilai z = {x[2]:.4f}")
```
